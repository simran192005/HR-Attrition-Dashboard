```python
import pandas as pd 

df= pd.read_csv(r'C:\Users\ASUS\OneDrive\Documents\HR-Attrition-Dashboard\SampleData_-HR-Employee-Attrition.csv')

print("Shape:",df.shape)
print("\nNull Values:\n",df.isnull().sum())
print("\nAttrition Count:\n", df['Attrition'].value_counts())
```

```python
df= df.drop(columns=['EmployeeCount', 'Over18', 'StandardHours'])
df['AttritionFlag']= df['Attrition'].map({'Yes':1, 'No':0})
df.to_csv('HR_Attrition_Clean.csv',index=False)
```

```python
print("Clean data shape:", df.shape)
print("\nColumns remaining:", list(df.columns))
print("\nAttritionFlag check:\n", df[['Attrition', 'AttritionFlag']].head(5))
```

## Phase 1 - Data Cleaning
- Dataset: IBM HR Analytics (1470 rows, 35 columns)
- No null values found
- Dropped 3 zero-variance columns: EmployeeCount, Over18, StandardHours
- Encoded Attrition column to numeric (Yes=1, No=0)
- Saved clean file as HR_Attrition_Clean.csv
